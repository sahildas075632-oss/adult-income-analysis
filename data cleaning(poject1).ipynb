```python
#take column name


columns = [
    "age",
    "workclass",
    "fnlwgt",
    "education",
    "education_num",
    "marital_status",
    "occupation",
    "relationship",
    "race",
    "sex",
    "capital_gain",
    "capital_loss",
    "hours_per_week",
    "native_country",
    "income"
]
```

```python
df_data = pd.read_csv(
    "/Users/sahildas/Downloads/adult 2/data/adult.data",
    header=None,
    names=columns
)
df_data
```

```python
#add new column

df_data["source_split"] = "train"
df_data
```

```python
df_test["source_split"] = "test"
df_test
```

```python
#combine two table

df = pd.concat([df_data, df_test], ignore_index=True)
df
```

```python
df.shape
```

```python
#remove the white space
df = df.apply(lambda x: x.str.strip() if x.dtype == "object" else x)
df
```

```python
#find howmany time question mark is prasent 
df.isin(["?"]).sum()
```

```python
#replace missing valu
df = df.replace("?", pd.NA)
df
```

```python
#find missing value
df.isna().sum()
```

```python
#convert the  issing value with unknown

df["workclass"] = df["workclass"].fillna("Unknown")
df["workclass"]
```

```python
df["occupation"] = df["occupation"].fillna("Unknown")
df["occupation"]
```

```python
df["native_country"] = df["native_country"].fillna("Unknown")
df["native_country"]
```

```python
df.isna().sum()
```

```python
#check the numeric values 
df.describe()
```

```python
#chack min value
df[["age", "education_num", "hours_per_week"]].min()
```

```python
#chack max valu
df[["age", "education_num", "hours_per_week"]].max()
```

```python
#chack duplicate row
df.duplicated().sum()
```

```python
#chack unique value
df["income"].unique()
```

```python
#remove the dot
df["income"] = df["income"].str.replace(".", "", regex=False)
df["income"]
```

```python
df["income"].unique()
```

```python
#convert hig and low incom
df["income"] = df["income"].map({
    "<=50K": "low_income",
    ">50K": "high_income"
})
df["income"]
```

```python
#count low and high income
df["income"].value_counts()
```

```python
#chack age column
df["age"].describe()
```

```python
#creat a new column age_band.
#pd.cut() is a Pandas function used to divide numerical values into groups/ranges.
#bins means the boundaries used to divide numbers into groups.
#labels means the names we give to each group created by bins

df["age_band"] = pd.cut(df["age"], bins=[16, 25, 35, 45, 55, 65, 90], labels=["17-25", "26-35", "36-45", "46-55", "56-65", "66-90"])
df["age_band"]
```

```python
#count the  diffarent age_band

df["age_band"].value_counts()
```

```python
df
```

```python
#chack about the hours_per_week column

df["hours_per_week"].describe()
```

```python
#creat hoursbond column
df["hours_band"] = pd.cut(df["hours_per_week"], bins=[0, 20, 30, 40, 50, 60, 99], labels=["1-20", "21-30", "31-40", "41-50", "51-60", "61-99"])
df["hours_band"]
```

```python
df
```

```python
##count the  diffarent hours_band 
df["hours_band"].value_counts()
```

```python
#creat a new column net_capital inside it assienged the calculated value

df["net_capital"] = df["capital_gain"] - df["capital_loss"]
df["net_capital"]
```

                                                              #EDA(Exploratory Data Analysis)                                                         

```python
#what percentage of people belong to each incom group

df["income"].value_counts(normalize=True) * 100
```

```python
#compare high-income percentage by education.

df.groupby("education")["income"].apply(lambda x: (x == "high_income").mean() * 100)
```

```python
#bar chat

education_income = df.groupby("education")["income"].apply(lambda x: (x == "high_income").mean() * 100)
education_income.sort_values().plot(kind="barh", figsize=(10, 6))
plt.xlabel("High-income percentage")
plt.ylabel("Education")
plt.title("High-Income Percentage by Education")
plt.show()
```

```python
#high incom porcenteg

occupation_income = df.groupby("occupation")["income"].apply(
    lambda x: (x == "high_income").mean() * 100
)

occupation_income
```

```python
#Occupation bar chart

occupation_income.sort_values().plot(
    kind="barh",
    figsize=(10, 6)
)

plt.xlabel("High-income percentage")
plt.ylabel("Occupation")
plt.title("High-Income Percentage by Occupation")
plt.show()
```

```python
#Calculate high-income percentage by age band

age_income = df.groupby("age_band", observed=True)["income"].apply(
    lambda x: (x == "high_income").mean() * 100
)

age_income
```

```python
#bar chat

age_income.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.xlabel("Age Band")
plt.ylabel("High-income percentage")
plt.title("High-Income Percentage by Age Band")
plt.xticks(rotation=0)
plt.show()
```

```python
#calculates the percentage of high-income records within each workclass.

workclass_income = df.groupby("workclass")["income"].apply(
    lambda x: (x == "high_income").mean() * 100
)

workclass_income
```

```python
#Workclass bar chart.

workclass_income.sort_values().plot(
    kind="barh",
    figsize=(10, 6)
)

plt.xlabel("High-income percentage")
plt.ylabel("Workclass")
plt.title("High-Income Percentage by Workclass")
plt.show()
```

```python
#calculates the percentage of high-income records within each sex category.

sex_income = df.groupby("sex")["income"].apply(
    lambda x: (x == "high_income").mean() * 100
)

sex_income
```

```python
#Make the Sex bar chart

sex_income.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.xlabel("Sex")
plt.ylabel("High-income percentage")
plt.title("High-Income Percentage by Sex")
plt.xticks(rotation=0)
plt.show()
```
